# Alpha Vantage — NEWS_SENTIMENT Test
Fetch and inspect the `NEWS_SENTIMENT` payload from Alpha Vantage for a single ticker.

```python
import json
import os

import requests
from dotenv import load_dotenv

from config import TICKERS
```

```python
# Load ALPHA_VANTAGE_API_KEY from the .env file in the project root
load_dotenv()
API_KEY = os.getenv("ALPHA_VANTAGE_API_KEY")

if not API_KEY:
    raise EnvironmentError(
        "ALPHA_VANTAGE_API_KEY not found. "
        "Make sure it is defined in your .env file."
    )

print("API key loaded successfully.")
```

```python
ticker = TICKERS[0]  # NVDA

url = "https://www.alphavantage.co/query"
params = {
    "function": "NEWS_SENTIMENT",
    "tickers": ticker,
    "apikey": API_KEY,
}

print(f"Target ticker : {ticker}")
print(f"Endpoint      : {url}")
```

```python
print(f"Fetching NEWS_SENTIMENT for {ticker} from Alpha Vantage...\n")

try:
    response = requests.get(url, params=params, timeout=15)
    response.raise_for_status()
    data = response.json()
    print(json.dumps(data, indent=2))
except requests.exceptions.HTTPError as e:
    print(f"HTTP error: {e} — Status code: {response.status_code}")
except requests.exceptions.ConnectionError as e:
    print(f"Connection error: {e}")
except requests.exceptions.Timeout:
    print("Request timed out. Alpha Vantage did not respond within 15 seconds.")
except requests.exceptions.RequestException as e:
    print(f"Unexpected request error: {e}")
```
